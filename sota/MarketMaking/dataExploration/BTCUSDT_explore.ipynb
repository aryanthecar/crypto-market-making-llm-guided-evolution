```python
# numba.njit is strongly recommended for fast backtesting.
@njit
def print_bbo(hbt):
    # Iterating until hftbacktest reaches the end of data.
    # Elapses 60-sec every iteration.
    # Time unit is the same as data's timestamp's unit.
    # Timestamp of the sample data is in nanoseconds.
    while hbt.elapse(60 * 1e9) == 0:        
        # Gets the market depth for the first asset.
        depth = hbt.depth(0)

        # Prints the best bid and the best offer.
        print(
            'current_timestamp:', hbt.current_timestamp,
            ', best_bid:', np.round(depth.best_bid, 1),
            ', best_ask:', np.round(depth.best_ask, 1)
        )
    return True
```

```python
import matplotlib.pyplot as plt
# plot ROI as neg_ROI and MaxDrawdown as points, if is_seed, color red, else color blue
plt.scatter(points['neg_ROI'], points['MaxDrawdown'], c=points['is_seed'].map({True: 'red', False: 'blue'}))
plt.xlabel('ROI')
plt.ylabel('MaxDrawdown')
plt.title('ROI vs MaxDrawdown')
plt.show()
```

```python
# create a pareto optimal front, where line starts at top of y axis and right of x axis
# plot all points, but connect the pareto optimal points via a line that goes down and to the right
# Get points as numpy arrays
x = points['neg_ROI'].values
y = points['MaxDrawdown'].values

# Find pareto optimal points
pareto_points = []
for i in range(len(x)):
    is_pareto = True
    for j in range(len(x)):
        if i != j:
            if x[j] >= x[i] and y[j] <= y[i]:
                is_pareto = False
                break
    if is_pareto:
        pareto_points.append((x[i], y[i]))

# Sort pareto points by x coordinate
pareto_points.sort()

# Plot original points
plt.scatter(points['neg_ROI'], points['MaxDrawdown'], c=points['is_seed'].map({True: 'red', False: 'blue'}))

# Plot pareto front line
pareto_x = [p[0] for p in pareto_points]
pareto_y = [p[1] for p in pareto_points]
plt.plot(pareto_x, pareto_y, 'g--', label='Pareto Front')

plt.xlabel('ROI')
plt.ylabel('MaxDrawdown')
plt.title('ROI vs MaxDrawdown with Pareto Front')
plt.legend()
plt.show()



```
